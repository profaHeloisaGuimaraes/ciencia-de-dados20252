```python
from sklearn.datasets import load_iris # Importe da base de dados

import pandas as pd
```

```python
# Ajustando o dataset -> Transformando em um Dataframe do Pandas
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler #Import do MinMaxScaler

scaler_mm = MinMaxScaler() #Criando uma instância do MinMaxScaler
df_minmax = scaler_mm.fit_transform( df[['sepal length (cm)']]) # Aplicando fit_transform() nos dados da coluna "sepal length (cm)" do DataFrame 'df'
```

```python
print("Min:", df_minmax.min()) # Apresentando o valor minimo da coluna
print("Max:", df_minmax.max()) # Apresentando o valor máximo da coluna
```

```python
# print(df_minmax) # Apresentando os valores da coluna normalizados
```

```python
from sklearn.preprocessing import StandardScaler # Importe do StandardScaler

scaler_ss = StandardScaler() #Criando uma instância do StandardScaler
df_standard = scaler_ss.fit_transform( df[['sepal length (cm)']])  # Aplicando fit_transform() nos dados da coluna "sepal length (cm)" do DataFrame 'df'
```

```python
print("Média:", df_standard.mean())
print("DP:", df_standard.std())
```

```python
print(df_standard)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 4))
plt.hist(df['sepal length (cm)'], label='Original', alpha=0.6)
plt.hist(df_minmax, label='MinMax', alpha=0.6)
plt.hist(df_standard, label='Standard', alpha=0.6)
plt.legend()
plt.show()
```
